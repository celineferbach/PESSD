## Part de la population de +65 ans

Calcule `pop_65 / pop_totale` pour chaque pays et chaque année.

```python
# Chargement
pop_65  = pd.read_excel('data/Pop+65ans_fr.xlsx',  index_col=0)
pop_tot = pd.read_excel('data/Pop_tot_fr.xlsx', index_col=0)

# Gestion des NaN
pop_65  = pop_65.replace(':', np.nan).astype(float)
pop_tot = pop_tot.replace(':', np.nan).astype(float)

# Calcul et exportation
part_65 = pop_65 / pop_tot
part_65.to_excel('data/part_pop_plus_65_fr.xlsx', index=True)
print('part_pop_plus_65_fr.xlsx exporté.')
```
